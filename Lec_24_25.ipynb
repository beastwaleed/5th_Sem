#Linear Regression vs Logistic Regression

Linear Regression is used to predict a quantity (how much?).

Logistic Regression is used to predict a probability or category (yes or no?).

**Linear Regression** (Continuous Output)
Imagine you are trying to predict the exact price of a house. The answer could be 150,000, 150,001, or 500,000. The output is a continuous number that can theoretically range from negative infinity to positive infinity (though context usually limits this).

Goal: Find a "line of best fit" that runs through your data points to minimize the error between the line and the actual values.

**Logistic Regression** (Categorical Output)
Imagine you are trying to predict if an email is Spam (1) or Not Spam (0). The answer isn't a number like "0.45 spam"; it's a classification. Logistic regression predicts the probability that an email is spam (e.g., "there is a 95% chance this is spam").

Goal: Find an "S-shaped curve" that separates the data into classes (0 vs 1)

#What is Logistic Regression

Logistic regression is a supervised machine learning algorithm used for classification problems where the goal is to predict a categorical outcome, such as "yes" or "no" or "spam" or "not spam". It works by modeling the probability of an event occurring using a logistic (or sigmoid) function, which maps any real-valued input to a value between 0 and 1. This output can be interpreted as the likelihood of an instance belonging to a particular class.

##Formula:


The logistic regression formula predicts the probability of a binary outcome using a two-step process: first, it calculates a linear combination of the input variables , and then it passes that result through the sigmoid function to produce a probability between 0 and 1.

##Interpreting Logistic Regression Coefficients

Interpreting the coefficients (weights/theta) in Logistic Regression is tricky because the relationship is not linear.

In Linear Regression ($y = 2x$), if $x$ goes up by 1, $y$ goes up by 2. Simple.In Logistic Regression, because of the S-curve, the effect of the weight changes depending on where you are on the curve.

image.png

Logistic Regression doesn't predict raw probability linearly; it predicts the Log Odds.$$\ln(\text{Odds}) = \theta_0 + \theta_1 x_1 + \dots$$If you increase $x$ by 1 unit, the "Log Odds" increase by $\theta$. To make this understandable, we take the exponential ($e^\theta$).The Rule: If you increase $x$ by 1 unit, the Odds of the outcome happen $e^\theta$ times more.

Real World Example:Imagine we are predicting if a user buys a product ($y=1$).

Feature: Time Spent on Site (minutes).

Coefficient ($\theta$): 0.693


$Calculation$:$$e^{0.693} \approx 2.0$$Interpretation:"For every extra minute a user spends on the site, their odds of buying the product double (increase by a factor of 2)."

`Important`
**Diff B/W Odds and Probability:**

**Probability**: The ratio of "Successes" to the Total outcomes.

    Question: "What is the chance I win?"

**Odds**: The ratio of "Successes" to Failures.

    Question: "How much stronger is my chance of winning compared to losing?
