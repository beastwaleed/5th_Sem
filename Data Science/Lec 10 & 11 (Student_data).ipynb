## Encoding Categorical Data
-----

```python
## create dataframe copy
df2 = df.copy()
df2
```

### 1. Label Encoder

```python
# import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Initialize LabelEncoder
label_encoder = LabelEncoder()

# Apply label encoding
df2['remarks'] = label_encoder.fit_transform(df2['remarks'])
df2
```

### 2. Ordinal Encoding

```python
print(df2.head())
df2 = df.copy()
print(df2.head)
```

```python
from sklearn.preprocessing import OrdinalEncoder

# Initialize OrdinalEncoder
ordinal_encoder = OrdinalEncoder(categories=[['Excellent', 'Good', 'Average', 'Poor']])

# Apply Ordinal Encoding
df2['remarks'] = ordinal_encoder.fit_transform(df2[['remarks']])

print("Ordinal Encoded Data:")
df2
```

### 3. One-Hot Encoding

```python
# Apply One-Hot Encoding
df_one_hot= pd.get_dummies(df2['remarks'], prefix='Remark')

print("One-Hot Encoded Data:")
df_one_hot 
```

### 4. Binary encoding

### 5. Frequency Encoding

```python
# # Sample data
# data = ['Apple', 'Banana', 'Apple', 'Orange', 'Banana']
# df = pd.DataFrame(data, columns=['Fruit'])

# Frequency Encoding
frequency_encoding = df2['remarks'].value_counts().to_dict()
print(frequency_encoding)
df2['Remarks_encoded'] = df2['remarks'].map(frequency_encoding)

print("Frequency Encoded Data:")
print(df2)
```
