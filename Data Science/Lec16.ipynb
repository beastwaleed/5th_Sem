### Problem 02

```python
import pandas as pd
import numpy as np
from scipy.stats import f_oneway

print("=== ONE-WAY ANOVA TEST ===")
print("Research Question: Do different teaching methods affect test scores?")

# Sample data: Test scores from three different teaching methods
method_A = [85, 88, 92, 78, 84, 90, 87]  # Traditional method
method_B = [92, 95, 89, 94, 91, 93, 90]  # Interactive method
method_C = [78, 82, 80, 85, 79, 81, 83]  # Online method

print(f"\nSAMPLE DATA:")
print(f"Method A (Traditional): {method_A}")
print(f"Method B (Interactive): {method_B}")
print(f"Method C (Online): {method_C}")

print(f"\nSAMPLE STATISTICS:")
print(f"Method A: Mean = {np.mean(method_A):.2f}, Std = {np.std(method_A, ddof=1):.2f}")
print(f"Method B: Mean = {np.mean(method_B):.2f}, Std = {np.std(method_B, ddof=1):.2f}")
print(f"Method C: Mean = {np.mean(method_C):.2f}, Std = {np.std(method_C, ddof=1):.2f}")

# Perform ANOVA test
f_stat, p_value = f_oneway(method_A, method_B, method_C)

print("\n" + "="*50)
print("HYPOTHESES:")
print("="*50)
print("Null Hypothesis (H₀): μ_A = μ_B = μ_C")
print("    (All teaching methods have equal mean test scores)")
print("Alternative Hypothesis (H₁): At least one μ differs")
print("    (At least one teaching method has different mean test score)")

print("\n" + "="*50)
print("ANOVA TEST RESULTS:")
print("="*50)
print(f"F-statistic: {f_stat:.4f}")
print(f"P-value: {p_value:.4f}")

print("\n" + "="*50)
print("INTERPRETATION:")
print("="*50)

if p_value < 0.05:
    print(f"✅ REJECT the null hypothesis (p = {p_value:.4f} < 0.05)")
    print("📊 CONCLUSION: There is a STATISTICALLY SIGNIFICANT difference")
    print("   in mean test scores between at least two teaching methods")
    print("💡 MEANING: The teaching method DOES affect test scores")
else:
    print(f"✅ FAIL TO REJECT the null hypothesis (p = {p_value:.4f} ≥ 0.05)")
    print("📊 CONCLUSION: There is NO statistically significant difference")
    print("   in mean test scores between the teaching methods")
    print("💡 MEANING: The teaching method does NOT affect test scores")

print("\n" + "="*50)
print("STATISTICAL DECISION:")
print("="*50)
print(f"Significance level (α): 0.05")
print(f"Decision: {'REJECT H₀' if p_value < 0.05 else 'FAIL TO REJECT H₀'}")
print(f"Evidence for difference: {'STRONG' if p_value < 0.01 else 'MODERATE' if p_value < 0.05 else 'WEAK'}")
```

### Problem 02

```python
import numpy as np
import pandas as pd
from scipy import stats
import scipy.stats as stats
from scipy.stats import chi2_contingency

# Test of Independence: Are gender and voting preference related?
print("=== CHI-SQUARE TEST OF INDEPENDENCE ===")

# Contingency table: Gender vs Voting Preference
observed_independence = np.array([
    [30, 20, 10],  # Male votes: [Democrat, Republican, Independent]
    [25, 15, 20]   # Female votes: [Democrat, Republican, Independent]
])

chi2_stat, p_value, dof, expected = chi2_contingency(observed_independence)

print("Contingency Table (Gender vs Voting Preference):")
print(observed_independence)
print(f"\nChi-square statistic: {chi2_stat:.4f}")
print(f"P-value: {p_value:.4f}")
print(f"Degrees of freedom: {dof}")
print(f"Significant relationship? {p_value < 0.05}")
```

```python
# Hypotheses
print("\nHYPOTHESES:")
print("Null (H₀): Gender and voting preference are INDEPENDENT")
print("Alternative (H₁): Gender and voting preference are RELATED")

# Interpretation
print("\nINTERPRETATION:")
if p_value < 0.05:
    print(f"✅ REJECT H₀ (p = {p_value:.4f} < 0.05)")
    print("There is a SIGNIFICANT relationship between gender and voting preference")
else:
    print(f"✅ FAIL TO REJECT H₀ (p = {p_value:.4f} ≥ 0.05)")
    print("There is NO SIGNIFICANT relationship between gender and voting preference")

print("\n" + "="*50)
print("SUMMARY: TEST OF INDEPENDENCE")
print("="*50)

# Independence Test
print("   - Tests: Are two variables related?")
print("   - Input: Contingency table (2D array)")
print("   - Function: chi2_contingency()")
print("   - H₀: Variables are independent")
```

### Problem 03

```python
print("=== CHI-SQUARE TEST OF INDEPENDENCE ===")
print("Research Question: Is there a relationship between gender and academic performance?")

# Sample students data
students_data = {
    'gender': ['Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female'],
    'performance': ['Good', 'Good', 'Average', 'Excellent', 'Average', 'Good', 'Excellent', 'Average']
}
df = pd.DataFrame(students_data)

# Create contingency table
contingency_table = pd.crosstab(df['gender'], df['performance'])
print("\nCONTINGENCY TABLE:")
print("(Rows: Gender, Columns: Performance)")
print(contingency_table)

# Perform chi-square test
chi2_stat, p_value, dof, expected = chi2_contingency(contingency_table)

print(f"\nEXPECTED FREQUENCIES (if no relationship):")
print(expected)

print("\nTEST RESULTS:")
print(f"Chi-square statistic: {chi2_stat:.4f}")
print(f"P-value: {p_value:.4f}")
print(f"Degrees of freedom: {dof}")

# STATE HYPOTHESES
print("\n" + "="*50)
print("HYPOTHESES:")
print("="*50)
print("Null Hypothesis (H₀): Gender and academic performance are INDEPENDENT")
print("Alternative Hypothesis (H₁): Gender and academic performance are RELATED")

# INTERPRETATION
print("\n" + "="*50)
print("INTERPRETATION:")
print("="*50)

if p_value < 0.05:
    print(f"✅ REJECT the null hypothesis (p = {p_value:.4f} < 0.05)")
    print("📊 CONCLUSION: There is a STATISTICALLY SIGNIFICANT relationship")
    print("   between gender and academic performance")
    print("💡 MEANING: Academic performance differs significantly between genders")
else:
    print(f"✅ FAIL TO REJECT the null hypothesis (p = {p_value:.4f} ≥ 0.05)")
    print("📊 CONCLUSION: There is NO statistically significant relationship")
    print("   between gender and academic performance")
    print("💡 MEANING: Academic performance is similar across genders")

# ADDITIONAL INSIGHTS
print("\n" + "="*50)
print("ADDITIONAL INSIGHTS:")
print("="*50)
print(f"Significance level (α): 0.05")
print(f"Decision threshold: p-value {'<' if p_value < 0.05 else '≥'} 0.05")
print(f"Statistical significance: {'YES' if p_value < 0.05 else 'NO'}")
```

### Problem 05

```python
from scipy.stats import chisquare

# Goodness-of-Fit: Do dice rolls follow uniform distribution?
print("\n=== CHI-SQUARE GOODNESS-OF-FIT TEST ===")

# Observed dice rolls
observed_fit = np.array([18, 22, 15, 21, 19, 25])  # Rolls for faces 1-6
# Expected: uniform distribution (equal probability for each face)
expected_fit = np.array([20, 20, 20, 20, 20, 20])  # Expected if fair dice

chi2_stat, p_value = chisquare(observed_fit, expected_fit)

print("Observed dice rolls:", observed_fit)
print("Expected (uniform):  ", expected_fit)
print(f"\nChi-square statistic: {chi2_stat:.4f}")
print(f"P-value: {p_value:.4f}")
print(f"Follows expected distribution? {p_value >= 0.05}")

# Hypotheses
print("\nHYPOTHESES:")
print("Null (H₀): The dice is FAIR (follows uniform distribution)")
print("Alternative (H₁): The dice is BIASED (does not follow uniform distribution)")

# Interpretation
print("\nINTERPRETATION:")
if p_value < 0.05:
    print(f"✅ REJECT H₀ (p = {p_value:.4f} < 0.05)")
    print("The dice appears to be BIASED - distribution differs from expected")
else:
    print(f"✅ FAIL TO REJECT H₀ (p = {p_value:.4f} ≥ 0.05)")
    print("The dice appears to be FAIR - follows expected uniform distribution")

print("\n" + "="*50)
print("SUMMARY: GOODNESS-OF-FIT")
print("="*50)

# Goodness-of-Fit Test
print("   - Tests: Does data match expected distribution?")
print("   - Input: Observed counts vs expected counts (1D arrays)")
print("   - Function: chisquare()")
print("   - H₀: Data follows expected distribution")
```
