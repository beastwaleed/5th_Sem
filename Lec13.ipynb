### Binomial Distribution

```python
import pandas as pd
from scipy.stats import binom

# --- 1. Define Parameters ---
n_trials = 10     # Total number of coin flips
p_success = 0.5   # Probability of getting a Head (Success)
k_target = 5      # The specific number of Heads we are interested in

# --- 2. Calculate the PMF (Probability Mass Function) ---
# binom.pmf(k, n, p) calculates P(X=k)
prob_k_successes = binom.pmf(k=k_target, n=n_trials, p=p_success)

# --- 3. Create a DataFrame for a range of outcomes (0 to 10 heads) ---
# Possible outcomes: 0, 1, 2, ..., 10 heads
possible_heads = list(range(n_trials + 1))

# Calculate the probability for each possible outcome
all_probabilities = binom.pmf(possible_heads, n=n_trials, p=p_success)

# Create the DataFrame
df_binomial_pmf = pd.DataFrame({
    'Number_of_Heads (k)': possible_heads,
    'Probability P(X=k)': all_probabilities
})

# --- 4. Print Results ---
print(f"The probability of exactly {k_target} heads is: {prob_k_successes:.4f}\n")
print("--- Binomial Probability Mass Function (PMF) DataFrame ---")
print(df_binomial_pmf)
```

The Binomial distribution models the number of successes in a fixed number of independent trials, each with the same probability of success.

In this example, we'll simulate 10,000 trials of flipping a fair coin 10 times and count the number of heads in each set of 10 flips.

```python
import pandas as pd
from scipy.stats import binom

# --- Parameters for Binomial Distribution ---
n = 10     # Number of trials (e.g., coin flips)
p = 0.5    # Probability of success (e.g., getting a head)
size = 10000 # Number of simulations/samples to generate

# 1. Generate Random Variates (Samples)
# binom.rvs generates random variates (samples)
binomial_samples = binom.rvs(n=n, p=p, size=size)

# 2. Create a Pandas DataFrame
df_binomial = pd.DataFrame({
    'Trial_ID': range(1, size + 1),
    'Number_of_Successes': binomial_samples
})

# 3. Optional: Calculate the Probability Mass Function (PMF)
# PMF gives the probability of getting exactly 'k' successes.
# Here we calculate the probability for possible outcomes (0 to n)
k_values = range(n + 1)
pmf_values = binom.pmf(k_values, n=n, p=p)

# 4. Add PMF to the DataFrame (for the theoretical distribution)
df_pmf_binomial = pd.DataFrame({
    'k_Successes': k_values,
    'Probability_PMF': pmf_values
})

print("--- Binomial Samples DataFrame (First 5 Rows) ---")
print(df_binomial.head())
print("\n--- Binomial PMF DataFrame ---")
print(df_pmf_binomial)
```

### Poisson Distribution

image.png

In Poisson, **Mean = λ = Variance**

**p < 0.1 & n > 100**

```python
import pandas as pd
from scipy.stats import poisson

# --- 1. Define Parameters ---
lambda_avg = 3    # Average number of calls per hour (mu in scipy)
k_target = 2      # The specific number of calls we are interested in

# --- 2. Calculate the PMF (Probability Mass Function) ---
# poisson.pmf(k, mu) calculates P(X=k)
prob_k_events = poisson.pmf(k=k_target, mu=lambda_avg)

# --- 3. Create a DataFrame for a range of outcomes (0 to 8 calls) ---
# It's a discrete distribution, but we stop at a reasonable number (e.g., 8)
possible_calls = list(range(9))

# Calculate the probability for each possible outcome
all_probabilities = poisson.pmf(possible_calls, mu=lambda_avg)

# Create the DataFrame
df_poisson_pmf = pd.DataFrame({
    'Number_of_Calls (k)': possible_calls,
    'Probability P(X=k)': all_probabilities
})

# --- 4. Print Results ---
print(f"The probability of exactly {k_target} calls in an hour is: {prob_k_events:.4f}\n")
print("--- Poisson Probability Mass Function (PMF) DataFrame ---")
print(df_poisson_pmf)
```

### Normal Distribution

The Normal Distribution, also known as the Gaussian Distribution or bell curve, is the most important continuous probability distribution in statistics. It describes how the values of a variable are distributed, with most observations clustering around a central region and values tapering off symmetrically as they move away from the center
